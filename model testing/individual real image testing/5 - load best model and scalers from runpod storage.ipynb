Cell 3 — create local folders

Cell 4 — create the S3 client

```python
# Create the S3-compatible client using RunPod credentials
s3 = boto3.client(
    "s3",                                  # Use the S3 service
    aws_access_key_id=ACCESS_KEY,          # Pass the access key
    aws_secret_access_key=SECRET_KEY,      # Pass the secret key
    endpoint_url=ENDPOINT_URL,             # Pass the RunPod endpoint
    region_name=REGION,                    # Pass the region
)

# Print connection setup status
print("✅ S3 client created successfully.")
```

Cell 5 — optionally test bucket access

```python
# Print start message
print("🔍 Testing bucket access...")

# Request the list of objects in the bucket
response = s3.list_objects_v2(Bucket=BUCKET)

# Print whether objects were found
if "Contents" in response:
    print(f"✅ Connected to bucket '{BUCKET}' successfully.")
    print(f"Found {len(response['Contents'])} objects in this response page.")
else:
    print(f"⚠️ Connected to bucket '{BUCKET}', but no objects were returned.")
```
