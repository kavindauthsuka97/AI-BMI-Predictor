```python
# Import pickle to load saved feature data
import pickle

# Import numpy for array manipulation
import numpy as np

# Import pandas to create DataFrame
import pandas as pd

# Print confirmation
print("Step 1 completed: Libraries imported successfully.")
```

Cell 2 – Load the pickle file

```python
# Define the pickle file path
pickle_path = "front_mask_feature.pkl"

# Print status message
print(f"Loading feature file from: {pickle_path}")

# Open the pickle file in read-binary mode
with open(pickle_path, "rb") as f:
    feature_data = pickle.load(f)

# Print confirmation
print("Step 2 completed: Feature file loaded successfully.")

# Print type of loaded data
print("Type of loaded data:", type(feature_data))
```

Cell 3 – Inspect one element (understand shape)

```python
# Print status message
print("Inspecting the loaded feature data...")

# Print full shape of the feature data
print("Feature data shape:", feature_data.shape)

# Print first few values
print("First 10 feature values:")
print(feature_data[0][:10])

# Print confirmation
print("Step 3 completed: Feature data inspected successfully.")
```

Cell 4 – Convert to 2D array

```python
# Print status message
print("Converting feature data into 2D array...")

# Ensure the feature data is a NumPy array
features_matrix = np.array(feature_data)

# Print shape of the matrix
print("Shape of features_matrix:", features_matrix.shape)

# Confirm it is 2D
print("Number of dimensions:", features_matrix.ndim)

# Print confirmation
print("Step 4 completed: Feature data converted to 2D array.")
```

Cell 5 – Create DataFrame

```python
# Print status message
print("Creating DataFrame from feature matrix...")

# Get number of features (columns)
num_features = features_matrix.shape[1]

# Create column names: f1, f2, ..., fN
feature_columns = [f"f{i+1}" for i in range(num_features)]

# Create the DataFrame
eff_front_train = pd.DataFrame(features_matrix, columns=feature_columns)

# Print confirmation
print("Step 5 completed: DataFrame created successfully.")

# Print shape of DataFrame
print(f"DataFrame shape (rows x cols): {eff_front_train.shape}")

# Print first few column names
print("First 10 column names:", eff_front_train.columns.tolist()[:10])

# Preview the DataFrame
print("Preview of DataFrame:")
print(eff_front_train.head())
```

## Create dataframe for side mask

Cell 1 – Import required libraries

```python
# Import pickle to load saved feature data
import pickle

# Import numpy for array manipulation
import numpy as np

# Import pandas to create DataFrame
import pandas as pd

# Print confirmation
print("Step 1 completed: Libraries imported successfully.")
```

Cell 2 – Load the pickle file

```python
# Define the pickle file path
pickle_path = "side_mask_feature.pkl"

# Print status message
print(f"Loading feature file from: {pickle_path}")

# Open the pickle file in read-binary mode
with open(pickle_path, "rb") as f:
    feature_data = pickle.load(f)

# Print confirmation
print("Step 2 completed: Feature file loaded successfully.")

# Print type of loaded data
print("Type of loaded data:", type(feature_data))
```

Cell 3 – Inspect one element (understand shape)

```python
# Print status message
print("Inspecting the loaded feature data...")

# Print full shape of the feature data
print("Feature data shape:", feature_data.shape)

# Print first few values
print("First 10 feature values:")
print(feature_data[0][:10])

# Print confirmation
print("Step 3 completed: Feature data inspected successfully.")
```

Cell 4 – Convert to 2D array

```python
# Print status message
print("Converting feature data into 2D array...")

# Ensure the feature data is a NumPy array
features_matrix = np.array(feature_data)

# Print shape of the matrix
print("Shape of features_matrix:", features_matrix.shape)

# Print number of dimensions
print("Number of dimensions:", features_matrix.ndim)

# Print confirmation
print("Step 4 completed: Feature data converted to 2D array.")
```

Cell 5 – Create DataFrame

```python
# Print status message
print("Creating DataFrame from side mask feature matrix...")

# Get number of features (columns)
num_features = features_matrix.shape[1]

# Create column names: s1, s2, ..., sN
feature_columns = [f"s{i+1}" for i in range(num_features)]

# Create the DataFrame
eff_side_train = pd.DataFrame(features_matrix, columns=feature_columns)

# Print confirmation
print("Step 5 completed: DataFrame created successfully.")

# Print shape of DataFrame
print(f"DataFrame shape (rows x cols): {eff_side_train.shape}")

# Print first few column names
print("First 10 column names:", eff_side_train.columns.tolist()[:10])

# Preview the DataFrame
print("Preview of DataFrame:")
print(eff_side_train.head())
```

## Merge front mask and side mask datasets

```python
# Print status message
print("Merging front and side feature DataFrames...")

# Merge horizontally (axis=1 means column-wise)
combined_df = pd.concat([eff_front_train, eff_side_train], axis=1)

# Print confirmation
print("Step 2 completed: DataFrames merged successfully.")
```

```python
# Print shape of merged DataFrame
print("Combined DataFrame shape:", combined_df.shape)

# Print first few column names
print("First 10 columns:", combined_df.columns.tolist()[:10])

# Print last few column names
print("Last 10 columns:", combined_df.columns.tolist()[-10:])

# Preview the merged DataFrame
print("Preview of combined DataFrame:")
print(combined_df.head())

# Confirm completion
print("Step 3 completed: Merged DataFrame verified successfully.")
```

## Add height_cm and gender

```python
# Print status message
print("Adding gender feature...")

# Add new column 'gender' with value 'male'
combined_df["gender"] = "male"

# Print confirmation
print("Gender feature added successfully.")

# Verify
print("gender column value:")
print(combined_df["gender"])

# Confirm step
print("Step 3 completed: Gender feature added.")
```

```python
# Print status message
print("Adding height feature...")

# Add new column 'height_cm' with value 175
combined_df["height_cm"] = 175

# Print confirmation
print("Height feature added successfully.")

# Verify
print("height_cm column value:")
print(combined_df["height_cm"])

# Confirm step
print("Step 2 completed: Height feature added.")
```

```python
# Print final shape
print("Final DataFrame shape:", combined_df.shape)

# Print last few columns to confirm additions
print("Last 5 columns:", combined_df.columns.tolist()[-5:])

# Preview updated DataFrame
print("Preview after adding new features:")
print(combined_df.head())

# Confirm step
print("Step 4 completed: Final DataFrame verified.")
```

## Data preprocessing

```python
# Import pandas (if not already imported)
import pandas as pd

# Print status message
print("Converting 'gender' column to categorical and then to numeric codes...")

# Convert 'gender' column to categorical type
combined_df['gender'] = combined_df['gender'].astype('category')

# Convert categorical values to numeric codes
combined_df['gender'] = combined_df['gender'].cat.codes

# Print confirmation
print("Conversion completed successfully.")

# Show updated values
print("Updated 'gender' column:")
print(combined_df['gender'])

# Print data type
print("Data type of 'gender':", combined_df['gender'].dtype)

# Confirm step completion
print("Step completed: 'gender' converted to numeric codes.")
```
