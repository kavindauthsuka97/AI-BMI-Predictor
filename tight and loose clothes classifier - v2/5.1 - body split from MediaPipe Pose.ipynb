```python
from ultralytics import YOLO          # Import YOLOv8
import cv2                            # Import OpenCV
import numpy as np                    # Import NumPy

# Load YOLOv8 pose model (pretrained)
model = YOLO("yolov8n-pose.pt")

# Read input image
image = cv2.imread("man.png")

# Get image height
height, width, _ = image.shape

# Run pose detection
results = model(image)

# Get keypoints for first detected person
keypoints = results[0].keypoints.xy[0].cpu().numpy()

# COCO format indices for hips
LEFT_HIP = 11
RIGHT_HIP = 12

# Get hip y-coordinates
left_hip_y = keypoints[LEFT_HIP][1]
right_hip_y = keypoints[RIGHT_HIP][1]

# Compute waist line
waist_y = int((left_hip_y + right_hip_y) / 2)

# Split image into upper and lower body
upper_body = image[:waist_y, :]
lower_body = image[waist_y:, :]

# Save output images
cv2.imwrite("upper.png", upper_body)
cv2.imwrite("lower.png", lower_body)

print("Saved upper.png and lower.png")
```
