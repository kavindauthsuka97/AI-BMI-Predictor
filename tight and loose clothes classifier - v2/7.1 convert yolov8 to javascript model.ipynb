```python
from ultralytics import YOLO

model = YOLO("yolov8n-pose.pt")

# Export to TFJS. This will create a folder like: yolov8n-pose_web_model/
# nms=True bakes in NMS to reduce the amount of postprocessing you need in JS.
export_dir = model.export(format="tfjs", imgsz=640, batch=1, nms=True)

print("✅ Exported TFJS model to:", export_dir)
```

```python
import os, glob

# Ultralytics returns the folder path
export_dir = str(export_dir)

model_json = os.path.join(export_dir, "model.json")
shards = glob.glob(os.path.join(export_dir, "*.bin"))

print("model.json:", model_json, "exists =", os.path.isfile(model_json))
print("num weight shards:", len(shards))
print("example shards:", shards[:3])
```
