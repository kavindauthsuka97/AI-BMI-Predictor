```python
import os
from huggingface_hub import snapshot_download

def download_susy_model(local_dir="./models/susy", repo_id="HPAI-BSC/SuSy"):
    """
    Downloads SuSy.pt into local_dir and returns the local path.
    Will NOT re-download if it already exists in that folder.
    """
    os.makedirs(local_dir, exist_ok=True)
    local_pt = os.path.join(local_dir, "SuSy.pt")

    if os.path.exists(local_pt):
        print(f"[OK] Model already exists locally: {local_pt}")
        return local_pt

    print(f"[1] Downloading SuSy model from HF repo: {repo_id}")
    print(f"[2] Saving to local directory: {os.path.abspath(local_dir)}")

    # This downloads the repo snapshot but only keeps the SuSy.pt file
    snapshot_download(
        repo_id=repo_id,
        local_dir=local_dir,
        local_dir_use_symlinks=False,   # ensures a real file is copied into local_dir
        allow_patterns=["SuSy.pt"],     # download only the model file
    )

    if not os.path.exists(local_pt):
        raise FileNotFoundError(f"[ERR] Download completed but SuSy.pt not found at: {local_pt}")

    print(f"[OK] Download complete: {local_pt}")
    return local_pt


if __name__ == "__main__":
    model_path = download_susy_model("./models/susy")
    print("Saved model path:", model_path)
```
